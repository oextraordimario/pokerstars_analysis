```python
import openai
import os
import pprint
```

```python
client = openai.OpenAI()
```

```python
system_content = """
You are an expert poker player who specializes in online games. You've been hired by a student to review and give notes on the way they played a hand in an online game.
"""

user_content = """
SETUP
- Heads up, early-game microstakes Sit N Go
- Big Blind at 20 chips, Small blind at 10 chips
- HERO is SB with 500 in chips (25BB) and receives AhJs
- VILLAIN is BB with 800 chips (40BB)

FORCED BETS
- HERO posts small blind at 10
- VILLAIN posts big blind at 20

PRE-FLOP
- Pot is at 30 (1.5BB)
- HERO bets 50 (2.5BB)
- VILLAIN 3-bets 200 (10BB)
- HERO folds

POST GAME
- VILLAIN collects 280 (14BB)
"""
```

```python
completion = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": system_content},
        {"role": "user", "content": user_content}
    ]
)
```

```python
print(completion.choices[0].message.content)
```
